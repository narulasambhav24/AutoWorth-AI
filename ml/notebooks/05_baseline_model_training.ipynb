# AutoWorth-AI — Baseline Model Training

This notebook establishes baseline regression models for used-car price prediction.

## Objectives

- Establish a simple baseline using Dummy Regressor
- Train Linear Regression and Ridge Regression models
- Train a Random Forest baseline
- Evaluate models using MAE, RMSE, and R²
- Compare model performance
- Identify a strong baseline for further improvement

## Evaluation Strategy

Models will initially be evaluated using the train/test split established during preprocessing.

A grouped evaluation strategy will later be used to test generalization when identical feature profiles are prevented from appearing in both training and test sets.

```python
import pandas as pd
import numpy as np

from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression, Ridge
from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
DATA_PATH = "../data/processed/cardekho_model_ready.csv"

df = pd.read_csv(DATA_PATH)

df.shape
```

```python
df.head()
```

```python
df.info()
```

```python
df.columns.tolist()
```

Separate features and target

```python
TARGET = "selling_price"

X = df.drop(columns=TARGET)
y = df[TARGET].copy()
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
print("Target in X:", TARGET in X.columns)
```

```python
df.shape
```

```python
print("X shape:", X.shape)
print("y shape:", y.shape)
print("Target in X:", TARGET in X.columns)
```

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
print("X_train:", X_train.shape)
print("X_test:", X_test.shape)
print("y_train:", y_train.shape)
print("y_test:", y_test.shape)
```

Creating the baseline Model

```python
dummy_model = DummyRegressor(strategy="mean")
```

```python
dummy_model.fit(X_train, y_train)
```

Make Predictions

```python
y_pred_dummy = dummy_model.predict(X_test)
```

```python
print("Predictions shape:", y_pred_dummy.shape)
print("First 5 predictions:", y_pred_dummy[:5])
```

```python
dummy_model = DummyRegressor(strategy="mean")
```

```python
dummy_model.fit(X_train, y_train)
```

```python
y_pred_dummy = dummy_model.predict(X_test)

print("Predictions shape:", y_pred_dummy.shape)
print("First 5 predictions:", y_pred_dummy[:5])
```

Calculating the metrics

```python
dummy_mae = mean_absolute_error(y_test, y_pred_dummy)
dummy_rmse = np.sqrt(mean_squared_error(y_test, y_pred_dummy))
dummy_r2 = r2_score(y_test, y_pred_dummy)

print("Dummy Regressor")
print("MAE :", dummy_mae)
print("RMSE:", dummy_rmse)
print("R²  :", dummy_r2)
```

## Linear Regression

```python
import joblib

MODEL_PATH = "../models/preprocessor.joblib"

preprocessor = joblib.load(MODEL_PATH)
```

```python
X_train_transformed = preprocessor.transform(X_train)
X_test_transformed = preprocessor.transform(X_test)
```

```python
print("X_train transformed:", X_train_transformed.shape)
print("X_test transformed :", X_test_transformed.shape)
```

```python
linear_model = LinearRegression()
```

```python
linear_model.fit(X_train_transformed, y_train)
```

Generate predictions

```python
y_pred_linear = linear_model.predict(X_test_transformed)

print("Predictions shape:", y_pred_linear.shape)
print("First 5 predictions:", y_pred_linear[:5])
```

Evaluate Linear Regression

```python
linear_mae = mean_absolute_error(y_test, y_pred_linear)
linear_rmse = np.sqrt(mean_squared_error(y_test, y_pred_linear))
linear_r2 = r2_score(y_test, y_pred_linear)

print("Linear Regression")
print("MAE :", linear_mae)
print("RMSE:", linear_rmse)
print("R²  :", linear_r2)
```

## Ridge Regression

Ridge Regression extends Linear Regression by adding L2 regularization.

It is useful when the dataset contains many features or correlated features, as it discourages excessively large coefficients and can improve generalization.

```python
ridge_model = Ridge(alpha=1.0)
```

```python
ridge_model.fit(X_train_transformed, y_train)
```

```python
y_pred_ridge = ridge_model.predict(X_test_transformed)

print("Predictions shape:", y_pred_ridge.shape)
print("First 5 predictions:", y_pred_ridge[:5])
```

## Evaluate Ridge Regression

```python
ridge_mae = mean_absolute_error(y_test, y_pred_ridge)
ridge_rmse = np.sqrt(mean_squared_error(y_test, y_pred_ridge))
ridge_r2 = r2_score(y_test, y_pred_ridge)

print("Ridge Regression")
print("MAE :", ridge_mae)
print("RMSE:", ridge_rmse)
print("R²  :", ridge_r2)
```

## Random Forest Regression

Random Forest is an ensemble of decision trees.

Unlike linear models, it can capture nonlinear relationships and interactions between features. It is therefore a useful next baseline for the used-car price prediction problem.

```python
rf_model = RandomForestRegressor(
    n_estimators=100,
    random_state=42,
    n_jobs=-1
)
```

```python
rf_model.fit(X_train_transformed, y_train)
```

```python
y_pred_rf = rf_model.predict(X_test_transformed)

print("Predictions shape:", y_pred_rf.shape)
print("First 5 predictions:", y_pred_rf[:5])
```

```python
rf_mae = mean_absolute_error(y_test, y_pred_rf)
rf_rmse = np.sqrt(mean_squared_error(y_test, y_pred_rf))
rf_r2 = r2_score(y_test, y_pred_rf)

print("Random Forest")
print("MAE :", rf_mae)
print("RMSE:", rf_rmse)
print("R²  :", rf_r2)
```

## Baseline Model Comparison — Random Split

```python
results_random = pd.DataFrame({
    "Model": [
        "Dummy Regressor",
        "Linear Regression",
        "Ridge Regression",
        "Random Forest"
    ],
    "MAE": [
        dummy_mae,
        linear_mae,
        ridge_mae,
        rf_mae
    ],
    "RMSE": [
        dummy_rmse,
        linear_rmse,
        ridge_rmse,
        rf_rmse
    ],
    "R2": [
        dummy_r2,
        linear_r2,
        ridge_r2,
        rf_r2
    ]
})

results_random
```

## Baseline Model Conclusion

Four regression models were evaluated using the same train/test split and the metrics MAE, RMSE, and R².

Random Forest achieved the best baseline performance, with an MAE of approximately ₹1.02 lakh, RMSE of approximately ₹5.22 lakh, and R² of 0.741.

It substantially outperformed the linear models, suggesting that nonlinear relationships and interactions between vehicle features are important for used-car price prediction.

The Dummy Regressor established a naive reference point, while Linear Regression and Ridge Regression provided interpretable linear baselines.

Further evaluation using grouped splitting and additional model tuning can be performed in a later phase.
